```python
import pandas as pd
```

```python
df = pd.read_csv("titanic.csv")
```

```python
durchschnittsalter = df["Age"].mean()
durchschnittsalter
```

```python
durchschnittsalter = df["Age"].mean().round(3)

print(durchschnittsalter)
```

```python
ueberlebt = df["Survived"].value_counts()
ueberlebt
```

```python
df["Status"] = df["Survived"].map({0: "gestorben", 1: "überlebt"})
df["Status"].value_counts()
```

```python
maenner = df[df["Sex"] == "male"]
```

```python
maenner["Status"].value_counts()
```

```python
frauen = df[df["Sex"] == "female"]
```

```python
frauen["Status"].value_counts()
```

```python
df.groupby("Sex")["Status"].value_counts()
```

```python
durchschnittpreis = df["Fare"].mean().round(2)
durchschnittpreis
```

```python
df.groupby("Pclass")["Fare"].mean().round(2)
```

```python
df["Embarked"].value_counts()
```

```python
df.pivot_table(index="Sex", columns=["Pclass", "Status"], values="Survived", aggfunc="size")
```
